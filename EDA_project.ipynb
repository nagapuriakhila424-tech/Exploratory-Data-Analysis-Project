```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

#STEP 2 : LOAD DATASET




```python
df = pd.read_csv(r"Dataset for EDA project.csv")
print("\nDATASET LOADED SUCCESSFULLY\n")
```

#STEP 8 : CHECK MISSING VALUES




```python
print("\nMISSING VALUES\n")

print(df.isnull().sum())
```

#STEP9: HANDLE MISSING VALUES


```python
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].mean()
)

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:

    df[col] = df[col].fillna(
        df[col].mode()[0]
    )

print("\nMISSING VALUES AFTER CLEANING\n")

print(df.isnull().sum())
```

#STEP 10  :CHECK DUPLICATE VALUES




```python
print("\nDUPLICATE VALUES\n")

print(df.duplicated().sum())
```

#STEP 11 : REMOVE DUPLICATES




```python
df = df.drop_duplicates()

print("\nDUPLICATES REMOVED SUCCESSFULLY\n")
```

#STEP 25 : OUTLIER DETECTION USING IQR


```python
print("\nOUTLIER DETECTION\n")

for col in numeric_df.columns:

    Q1 = numeric_df[col].quantile(0.25)

    Q3 = numeric_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - (1.5 * IQR)

    upper_bound = Q3 + (1.5 * IQR)
    outliers = numeric_df[
        (numeric_df[col] < lower_bound) |
        (numeric_df[col] > upper_bound)
    ]
    print(f"{col} : {len(outliers)} outliers")
```
